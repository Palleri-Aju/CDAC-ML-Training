# Logistic Regression — Hearing Test (Clean Classroom Notebook)

**Goal:** Predict whether a participant passes a high-frequency hearing test, using **age** and **physical score**. This is a **binary classification** problem: `0 = Fail`, `1 = Pass`.

**Dataset:** `hearing_test.csv` (5,000 participants). Place the CSV in the same folder as this notebook.  
**Workflow:** Load → Inspect → Split → Scale → Train → Interpret → Predict → Evaluate.  
**Plots:** Only one essential graph: the confusion matrix.

> This example shows statistical associations, not a medical diagnosis or proof of causation.

## 1. Import the required libraries
We use Pandas for the dataset, NumPy for calculations, Scikit-learn for preprocessing, training, and evaluation, and Matplotlib for **one** confusion-matrix plot.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, confusion_matrix, ConfusionMatrixDisplay,
    classification_report
)
```

## 2. Load and inspect the dataset
- `age`: Participant age in years.
- `physical_score`: Physical exam score.
- `test_result`: **0 = Fail**, **1 = Pass**.

We verify the row count, missing values, and class balance before modeling.

```python
df = pd.read_csv("hearing_test.csv")
print("Dataset shape:", df.shape)
display(df.head())

print("\nMissing values:")
print(df.isna().sum())
print("\nClass counts (0 = Fail, 1 = Pass):")
print(df["test_result"].value_counts().sort_index())
```

## 3. Select independent and dependent variables
**X** contains the inputs. **y** contains the correct class labels. We are predicting a category, **not** a continuous value.

```python
X = df[["age", "physical_score"]]
y = df["test_result"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## 4. Split into training and test data
We keep the source notebook's `test_size=0.10` and `random_state=101`, making the demonstration comparable to the original.

**Why split?** Fit the model on training rows, then evaluate predictions on rows it has not seen.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.10, random_state=101
)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])
```

## 5. Standardize the features
Age and physical score have different scales. `StandardScaler` transforms each input into a **z-score** using the training-set mean and standard deviation:

\[x_{scaled}=rac{x-\mu_{train}}{\sigma_{train}}\]

**Important:** `fit_transform()` learns scaling parameters on training rows only. `transform()` reuses them on test rows, avoiding data leakage.

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Training scaled shape:", X_train_scaled.shape)
print("Test scaled shape:", X_test_scaled.shape)
print("Training feature means (approximately 0):", X_train_scaled.mean(axis=0).round(3))
```

## 6. Train the Logistic Regression model
Despite its name, Logistic Regression is a **classification algorithm**. It first calculates a linear score:

\[z=b_0+b_1x_1+b_2x_2\]

and converts it into the probability of `test_result=1` using sigmoid:

\[P(	ext{Pass})=rac{1}{1+e^{-z}}\]

`fit()` learns coefficients and the intercept by optimizing a classification loss. `max_iter=1000` allows sufficient optimization iterations.

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled, y_train)

print("Classes:", model.classes_)
print("Model training complete.")
```

## 7. Interpret the fitted coefficients
- `intercept_[0]` is \(b_0\).
- `coef_[0]` contains \(b_1\) and \(b_2\).
- A **positive** coefficient increases the predicted *log-odds* of passing, keeping the other feature fixed; a **negative** coefficient decreases them.
- Because we standardized the inputs, each coefficient is per **one training standard deviation**, not per original unit.
- `exp(coefficient)` is the odds multiplier for a one-standard-deviation increase in that feature (holding the other feature fixed).

```python
coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient (log-odds)": model.coef_[0],
    "Odds multiplier (exp(coef))": np.exp(model.coef_[0])
})

print("Intercept:", round(model.intercept_[0], 4))
display(coefficients.round(4))
```

## 8. Understand one prediction mathematically
We'll take the **first test observation** and calculate the linear score \(z\) and sigmoid probability by hand using the fitted model coefficients.

Remember: the model was trained on **scaled** inputs, so the manual calculation must also use scaled inputs.

```python
sample = X_test.iloc[[0]]  # Double brackets keep a 2D DataFrame
sample_scaled = scaler.transform(sample)

b0 = model.intercept_[0]
b1, b2 = model.coef_[0]
x1, x2 = sample_scaled[0]

z = b0 + b1 * x1 + b2 * x2
prob_pass_manual = 1 / (1 + np.exp(-z))
prob_pass_sklearn = model.predict_proba(sample_scaled)[0, 1]

print("Original observation:")
display(sample)
print("Scaled inputs:", sample_scaled[0].round(4))
print(f"Linear score z = {z:.4f}")
print(f"Manual sigmoid probability of Pass = {prob_pass_manual:.4f}")
print(f"Scikit-learn probability of Pass = {prob_pass_sklearn:.4f}")
print("Predicted class:", model.predict(sample_scaled)[0])
print("Actual class:", y_test.iloc[0])
assert np.isclose(prob_pass_manual, prob_pass_sklearn)
```

## 9. Predict classes and probabilities for the test dataset
- `predict()` returns predicted **class labels** (0 or 1).
- `predict_proba()` returns class probabilities in the order `model.classes_` (here, `[0, 1]`).

At the default 0.5 decision threshold, probability of class 1 at least 0.5 gives class 1.

```python
y_pred = model.predict(X_test_scaled)
y_prob_pass = model.predict_proba(X_test_scaled)[:, 1]

results = pd.DataFrame({
    "Actual": y_test.to_numpy(),
    "Predicted": y_pred,
    "Probability_Pass": y_prob_pass
})
display(results.head(10).round(4))
```

## 10. Evaluate classification performance
- **Accuracy:** Fraction of predictions that are correct.
- **Precision (Pass):** Among predicted Pass cases, fraction that actually passed.
- **Recall (Pass):** Among actual Pass cases, fraction predicted Pass.
- **F1-score:** Harmonic mean of precision and recall.

These measurements answer different questions, so accuracy alone is not always sufficient.

```python
print("Test accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("\nClassification report:\n")
print(classification_report(y_test, y_pred, target_names=["Fail (0)", "Pass (1)"]))

cm = confusion_matrix(y_test, y_pred, labels=[0, 1])
print("Confusion matrix (rows = actual, columns = predicted):")
print(cm)
```

## 11. The only required plot: Confusion Matrix
This visual shows correct and incorrect classifications. The diagonal cells are correct predictions; off-diagonal cells are errors.

```python
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Fail (0)", "Pass (1)"]
)
fig, ax = plt.subplots(figsize=(5.5, 4.5))
disp.plot(ax=ax, cmap="Blues", values_format="d", colorbar=False)
ax.set_title("Logistic Regression — Test Confusion Matrix")
plt.tight_layout()
plt.show()
```

## 12. Predict for a new participant
To obtain a valid prediction, apply **the same scaler trained on the training data** to the new observation.

These values illustrate the model workflow, not a medical assessment.

```python
new_person = pd.DataFrame({
    "age": [45.0],
    "physical_score": [35.0]
})

new_person_scaled = scaler.transform(new_person)
new_probability = model.predict_proba(new_person_scaled)[0, 1]
new_class = int(model.predict(new_person_scaled)[0])

print("Predicted probability of Pass:", round(new_probability, 4))
print("Predicted class:", new_class, "(1 = Pass, 0 = Fail)")
```
